```python
# Задание 1 — Работа со строками

s = input("Введите строку: ")

print("Длина строки:", len(s))
print("Верхний регистр:", s.upper())
print("Нижний регистр:", s.lower())

if len(s) >= 3:
    print("Первые 3 символа:", s[:3])
    print("Последние 3 символа:", s[-3:])
else:
    print("Строка слишком короткая для вывода первых и последних 3 символов.")
```

```python
# Задание 2 — Преобразование типов

a = int(input("Введите первое число: "))
b = int(input("Введите второе число: "))

print("Сумма:", a + b)
print("Разность:", a - b)
print("Произведение:", a * b)
```

```python
# Задание 3 — Поиск чисел с помощью регулярных выражений

import re

text = input("Введите текст: ")
numbers = re.findall(r"\d+", text)

print("Найденные числа:", numbers)
```

```python
# Задание 4 — Проверка номера телефона

import re

def check_phone(number: str) -> bool:
    pattern = r"^(\+7-\d{3}-\d{3}-\d{2}-\d{2}|8\(\d{3}\)\d{3}-\d{2}-\d{2})$"
    return bool(re.match(pattern, number))

# Примеры
tests = [
    "+7-999-123-45-67",
    "8(999)123-45-67",
    "8(000)000-00-00",
    "12345",
    "7-999-123-45-67"
]

for t in tests:
    print(t, "—", check_phone(t))
```

```python
# Задание 5 — Валидация пароля

import re

def check_password(password: str) -> bool:
    pattern = r"^(?=.*[A-Z])(?=.*[a-z])(?=.*\d).{8,}$"
    return bool(re.match(pattern, password))

# Примеры
tests = [
    "Password1",
    "badpass",
    "NOLOWER123",
    "lower1234",
    "ValidPass9"
]

for t in tests:
    print(t, "—", check_password(t))
```
