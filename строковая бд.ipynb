```python
# ЗАДАНИЕ 1: структура данных студентов

db = []  # база студентов в памяти

def create_student(last_name: str, first_name: str, group: str, gpa: float, year: int = None, record_book: str = None):
    """Добавить студента в базу"""
    student = {
        'last_name': last_name,
        'first_name': first_name,
        'group': group,
        'gpa': float(gpa),
        'year': year,
        'record_book': record_book
    }
    db.append(student)
    return student

def read_all():
    """Вернуть список всех студентов"""
    return db
```

```python
# ЗАДАНИЕ 2: поиск по GPA

def find_by_gpa(min_gpa: float):
    """Найти студентов со средним баллом выше min_gpa"""
    return [s for s in db if s.get('gpa', 0) > min_gpa]

# Пример использования
top_students = find_by_gpa(4.0)
print("Студенты с GPA > 4.0:", top_students)
```

```python
# ЗАДАНИЕ 3: удаление студентов по группе

def delete_by_group(group: str):
    """Удалить всех студентов указанной группы"""
    global db
    db = [s for s in db if s.get('group') != group]

# Пример использования
delete_by_group("ИВТ-101")
print("После удаления группы ИВТ-101:", db)
```

```python
# ЗАДАНИЕ 4: экспорт базы в CSV

import csv

def export_to_csv(filename: str):
    """Сохранить базу в CSV"""
    with open(filename, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=['last_name', 'first_name', 'group', 'gpa', 'year', 'record_book'])
        writer.writeheader()
        writer.writerows(db)

# Пример использования
export_to_csv('students.csv')
print("База экспортирована в students.csv")
```

```python
# ЗАДАНИЕ 5: поиск по регулярному выражению

import re

def find_by_regex(field: str, pattern: str):
    """Найти студентов, у которых поле field удовлетворяет regex pattern"""
    regex = re.compile(pattern)
    return [s for s in db if s.get(field) and regex.search(str(s[field]))]

# Пример использования
matching_students = find_by_regex('last_name', '^П')
print("Студенты, фамилия которых начинается на 'П':", matching_students)
```
