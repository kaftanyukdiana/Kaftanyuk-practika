```python
def power(base, exp):
    return base ** exp

power(2,3)
```

```python
def apply_to_list(func, items):
    return [func(x) for x in items]

apply_to_list(lambda x: x*2, [1,2,3])
```

```python
def make_adder(n):
    def adder(x):
        return x + n
    return adder

add5 = make_adder(5)
add5(10)
```

```python
import time

def timeit(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        print("Время:", time.time()-start)
        return result
    return wrapper

@timeit
def slow():
    for _ in range(10_0000):
        pass

slow()
```

```python
@timeit
def mixed(*args, **kwargs):
    print("args:", args, "kwargs:", kwargs)

mixed(1,2,a=3,b=4)
```
