```python
#Программа поверяет, содержится ли введеное слово в тексте
sentence = "На поляне расцвели красивые цветы" #Исходное предложение
word = input ("Введите слово для поиска:") #Пользовательно вводит слово

#Проверка наличия
if word in sentence:
    print ("Слово найдено!")
    print ("Позиция в тексте:", sentence.find(word)) #Показываем индекс
else:
    print("Такого слова нет в предложении.")
```

```python
#Программа берёт список чисел, составляет только нечетные и увеличивает каждое из них на 5

numbers = [3,10,7,8,1,4,9]

#filter() - выбираем нечетные
#map() - прибавляем 5
result = list(map(lambda x: x+5, filter(lambda x: x % 2! = 0, numbers)))

print("Исходный список:", numbers)
print("Результат после обработки:", result)
```

```python
#Создать список квадратов чисел от 1 до 10, но только для тех, которые делятся на 3

squares = [x**2 for x in range (1,11) if x % 3 == 0]

print("Квадраты чисел, кратных 3:", squares)
```

```python
products = [
    ("Куртка", "Одежда", 5000),
    ("Бананы", "Фрукты", 100),
    ("Джинсы", "Одежда", 4000),
    ("Груши", "Фрукты", 120)
    ]

#Применяем скидку 20% на одежду
discounted_clothes = list (
    map(
        lambda item: (item[0], item[2] * 0.8),
        filter(lambda item: item[1] == "Одежда",products)
    )
)
print(discounted_clothes)
```
